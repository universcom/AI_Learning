## 1. Basics & Figure Anatomy

```python
import matplotlib.pyplot as plt
import numpy as np

# ── Two styles of using Matplotlib ────────────────────────────────────────
# 1. pyplot (stateful, quick): plt.plot() ...
# 2. OO API  (explicit, recommended for complex plots): fig, ax = plt.subplots()
#
# Anatomy of a figure:
#   Figure  → the entire canvas (can hold multiple Axes)
#   Axes    → one plot area (has x-axis, y-axis, title, etc.)
#   Axis    → the actual x or y number line
#   Artist  → everything drawn: lines, text, patches

# ── Minimal figure with OO API ────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 4))   # figsize in inches

x = np.linspace(0, 2 * np.pi, 100)
ax.plot(x, np.sin(x))                   # draw a line

ax.set_title("My First Plot")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.grid(True)

plt.tight_layout()
plt.show()

# ── Key objects ────────────────────────────────────────────────────────────
print("Figure type :", type(fig))
print("Axes type   :", type(ax))
print("All axes    :", fig.axes)        # list of Axes in this Figure
```

## 2. Line Plots

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)

fig, ax = plt.subplots(figsize=(9, 4))

# ── color: name, hex, or (R,G,B) ─────────────────────────────────────────
ax.plot(x, np.sin(x),   color="steelblue",  label="sin(x)")
ax.plot(x, np.cos(x),   color="#e76f51",    label="cos(x)")
ax.plot(x, np.sin(2*x), color=(0.2,0.7,0.3),label="sin(2x)")

# ── linestyle options: '-'  '--'  '-.'  ':'  ─────────────────────────────
ax.plot(x, -np.sin(x),  linestyle="--", color="purple", label="-sin(x)")

# ── marker: 'o'  's'  '^'  'D'  'x'  '+'  ───────────────────────────────
ax.plot(x[::20], np.cos(2*x[::20]),
        marker="o", markersize=6, linestyle="none",
        color="tomato", label="cos(2x) samples")

# ── linewidth ─────────────────────────────────────────────────────────────
ax.plot(x, np.tan(np.clip(x, -1.5, 1.5)), linewidth=2.5,
        color="gray", alpha=0.5, label="tan(x) clipped")

ax.set_title("Line Plot — color, linestyle, marker")
ax.set_xlabel("x (radians)")
ax.set_ylabel("y")
ax.legend(loc="upper right")
ax.grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```

## 3. Scatter Plots

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)
n   = 150

x      = rng.normal(0, 1, n)
y      = 2 * x + rng.normal(0, 1, n)
colors = rng.uniform(0, 1, n)    # used as color intensity
sizes  = rng.uniform(20, 200, n) # point size in points²

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# ── Basic scatter ─────────────────────────────────────────────────────────
axes[0].scatter(x, y, alpha=0.7, color="steelblue", edgecolors="white", linewidths=0.5)
axes[0].set_title("Basic Scatter")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")

# ── Colored & sized by data values ────────────────────────────────────────
sc = axes[1].scatter(x, y,
                     c=colors,         # map a 3rd variable to color
                     s=sizes,          # map a 4th variable to size
                     cmap="viridis",   # colormap
                     alpha=0.8,
                     edgecolors="k",
                     linewidths=0.3)

fig.colorbar(sc, ax=axes[1], label="Color value")
axes[1].set_title("Scatter — color & size encode data")
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")

plt.tight_layout()
plt.show()
```

## 4. Bar Charts

```python
import matplotlib.pyplot as plt
import numpy as np

cities   = ["Tehran", "London", "Tokyo", "Paris", "NYC"]
sales_q1 = [42, 58, 61, 37, 75]
sales_q2 = [55, 63, 70, 45, 80]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# ── 1. Simple vertical bar ────────────────────────────────────────────────
axes[0].bar(cities, sales_q1, color="steelblue", edgecolor="white")
axes[0].set_title("Vertical Bar")
axes[0].set_ylabel("Sales (k)")
axes[0].set_xlabel("City")

# ── 2. Grouped bar ────────────────────────────────────────────────────────
x     = np.arange(len(cities))
width = 0.35

axes[1].bar(x - width/2, sales_q1, width, label="Q1", color="steelblue")
axes[1].bar(x + width/2, sales_q2, width, label="Q2", color="#e76f51")
axes[1].set_title("Grouped Bar")
axes[1].set_xticks(x)
axes[1].set_xticklabels(cities)
axes[1].legend()

# ── 3. Horizontal bar ─────────────────────────────────────────────────────
axes[2].barh(cities, sales_q1, color="mediumseagreen", edgecolor="white")
axes[2].set_title("Horizontal Bar")
axes[2].set_xlabel("Sales (k)")

# ── Add value labels on bars ──────────────────────────────────────────────
for rect, val in zip(axes[2].patches, sales_q1):
    axes[2].text(val + 0.5, rect.get_y() + rect.get_height()/2,
                 str(val), va="center", fontsize=9)

plt.tight_layout()
plt.show()
```

## 5. Histograms

```python
import matplotlib.pyplot as plt
import numpy as np

rng    = np.random.default_rng(42)
data_a = rng.normal(loc=60, scale=10, size=500)   # group A scores
data_b = rng.normal(loc=75, scale=8,  size=500)   # group B scores

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# ── 1. Basic histogram ────────────────────────────────────────────────────
axes[0].hist(data_a, bins=30, color="steelblue", edgecolor="white")
axes[0].set_title("Basic Histogram (bins=30)")
axes[0].set_xlabel("Score")
axes[0].set_ylabel("Count")

# ── 2. Overlapping histograms with alpha ──────────────────────────────────
axes[1].hist(data_a, bins=30, alpha=0.6, label="Group A", color="steelblue")
axes[1].hist(data_b, bins=30, alpha=0.6, label="Group B", color="#e76f51")
axes[1].set_title("Overlapping Histograms")
axes[1].set_xlabel("Score")
axes[1].legend()

# ── 3. Density (normalized) + cumulative ─────────────────────────────────
axes[2].hist(data_a, bins=30, density=True,  alpha=0.5,
             color="steelblue", label="PDF")
axes[2].hist(data_a, bins=30, density=True, cumulative=True,
             histtype="step", color="navy", linewidth=2, label="CDF")
axes[2].set_title("Density + Cumulative")
axes[2].set_xlabel("Score")
axes[2].set_ylabel("Probability")
axes[2].legend()

plt.tight_layout()
plt.show()
```

## 6. Pie Charts

```python
import matplotlib.pyplot as plt

labels  = ["Python", "JavaScript", "Java", "C++", "Other"]
sizes   = [38, 25, 17, 12, 8]
explode = [0.05, 0, 0, 0, 0]          # slightly pull out the largest slice
colors  = ["#4e9af1", "#f4a261", "#2a9d8f", "#e76f51", "#adb5bd"]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ── 1. Standard pie ───────────────────────────────────────────────────────
axes[0].pie(sizes,
            labels=labels,
            colors=colors,
            autopct="%1.1f%%",          # show percentage inside each slice
            explode=explode,
            shadow=True,
            startangle=140)             # rotate so largest slice is at top
axes[0].set_title("Language Popularity")

# ── 2. Donut chart (pie with a white circle in the centre) ────────────────
wedges, texts, autotexts = axes[1].pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct="%1.0f%%",
    startangle=140,
    wedgeprops=dict(width=0.55)         # donut width
)
# Style the percentage text
for at in autotexts:
    at.set_fontsize(9)
    at.set_color("white")

axes[1].set_title("Donut Chart")

plt.tight_layout()
plt.show()
```

## 7. Box Plots

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42)

# Simulate exam scores for 4 classes
data = [
    rng.normal(65, 12, 80),   # Class A
    rng.normal(72,  9, 80),   # Class B
    rng.normal(58, 15, 80),   # Class C — wider spread
    rng.normal(80,  6, 80),   # Class D — high performers
]
labels = ["Class A", "Class B", "Class C", "Class D"]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ── 1. Standard box plot ──────────────────────────────────────────────────
bp = axes[0].boxplot(data,
                     labels=labels,
                     patch_artist=True,    # fill boxes with color
                     notch=False)          # notch=True shows confidence interval

colors = ["#4e9af1", "#f4a261", "#2a9d8f", "#e76f51"]
for patch, color in zip(bp["boxes"], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)

axes[0].set_title("Box Plot — exam scores per class")
axes[0].set_ylabel("Score")
axes[0].grid(axis="y", linestyle=":", alpha=0.6)

# ── 2. Notched box plot + overlay raw data ────────────────────────────────
bp2 = axes[1].boxplot(data, labels=labels, patch_artist=True,
                      notch=True, showfliers=False)
for patch, color in zip(bp2["boxes"], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.5)

# Overlay jittered raw points
for i, (d, color) in enumerate(zip(data, colors), start=1):
    jitter = rng.uniform(-0.15, 0.15, len(d))
    axes[1].scatter(i + jitter, d, alpha=0.3, s=12, color=color)

axes[1].set_title("Notched + Raw Data Overlay")
axes[1].set_ylabel("Score")
axes[1].grid(axis="y", linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```

## 8. Subplots & Layouts

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)

# ── 1. subplots(): uniform grid ───────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(14, 6))
axes = axes.flatten()   # turn 2-D array into 1-D for easy iteration

funcs  = [np.sin, np.cos, np.tan, lambda x: np.sin(2*x),
          lambda x: np.cos(3*x), lambda x: x % (2*np.pi)]
titles = ["sin", "cos", "tan", "sin(2x)", "cos(3x)", "sawtooth"]

for ax, fn, title in zip(axes, funcs, titles):
    ax.plot(x, np.clip(fn(x), -3, 3), linewidth=1.5)
    ax.set_title(title)
    ax.grid(True, linestyle=":", alpha=0.5)

plt.suptitle("2×3 Subplot Grid", fontsize=14, y=1.01)
plt.tight_layout()
plt.show()

# ── 2. subplot2grid(): irregular grid ────────────────────────────────────
fig2 = plt.figure(figsize=(10, 5))

ax_big  = plt.subplot2grid((2, 3), (0, 0), colspan=2, rowspan=2)  # left large
ax_tr   = plt.subplot2grid((2, 3), (0, 2))                         # top-right
ax_br   = plt.subplot2grid((2, 3), (1, 2))                         # bot-right

ax_big.plot(x, np.sin(x), color="steelblue")
ax_big.set_title("Large panel (colspan=2, rowspan=2)")

ax_tr.plot(x, np.cos(x), color="#e76f51")
ax_tr.set_title("Top right")

ax_br.hist(np.sin(x), bins=15, color="mediumseagreen")
ax_br.set_title("Bottom right")

plt.suptitle("subplot2grid — irregular layout", fontsize=13)
plt.tight_layout()
plt.show()
```

## 9. Axes Customization

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(x, np.sin(x), color="steelblue", linewidth=2)

# ── Title & labels ────────────────────────────────────────────────────────
ax.set_title("Axes Customization Demo", fontsize=15, fontweight="bold", pad=12)
ax.set_xlabel("Angle (radians)", fontsize=12)
ax.set_ylabel("Amplitude",       fontsize=12)

# ── Axis limits ───────────────────────────────────────────────────────────
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-1.3, 1.3)

# ── Custom tick positions and labels ─────────────────────────────────────
ax.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
ax.set_xticklabels(["0", "π/2", "π", "3π/2", "2π"], fontsize=11)
ax.set_yticks([-1, -0.5, 0, 0.5, 1])

# ── Tick direction and size ───────────────────────────────────────────────
ax.tick_params(axis="both", direction="in", length=6, width=1.2)

# ── Grid: major and minor ─────────────────────────────────────────────────
ax.grid(which="major", linestyle="--", alpha=0.5)
ax.minorticks_on()
ax.grid(which="minor", linestyle=":", alpha=0.3)

# ── Hide top and right spines (clean look) ────────────────────────────────
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# ── Secondary y-axis ─────────────────────────────────────────────────────
ax2 = ax.twinx()
ax2.plot(x, np.cos(x), color="#e76f51", linewidth=1.5, linestyle="--", alpha=0.7)
ax2.set_ylabel("cos(x)", color="#e76f51", fontsize=12)
ax2.tick_params(axis="y", labelcolor="#e76f51")
ax2.spines["top"].set_visible(False)

plt.tight_layout()
plt.show()
```

## 10. Annotations & Text

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 300)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y, color="steelblue", linewidth=2)

# ── axhline / axvline: reference lines ───────────────────────────────────
ax.axhline(0,          color="gray",   linewidth=0.8, linestyle="--")
ax.axhline(1,          color="green",  linewidth=1,   linestyle=":",  label="max = 1")
ax.axhline(-1,         color="red",    linewidth=1,   linestyle=":",  label="min = -1")
ax.axvline(np.pi/2,    color="orange", linewidth=1,   linestyle="-.", label="π/2")

# ── text(): place text at data coordinates ────────────────────────────────
ax.text(np.pi/2 + 0.1, 0.85, "Peak\n(π/2, 1)", fontsize=10, color="darkgreen")

# ── annotate(): arrow + text pointing to a data point ────────────────────
ax.annotate("Zero crossing",
            xy=(np.pi, 0),              # point to annotate
            xytext=(np.pi + 0.5, 0.4), # where the label sits
            arrowprops=dict(arrowstyle="->", color="black", lw=1.5),
            fontsize=10)

ax.annotate("Trough",
            xy=(3*np.pi/2, -1),
            xytext=(3*np.pi/2 + 0.5, -0.5),
            arrowprops=dict(arrowstyle="->", color="red", lw=1.5),
            fontsize=10, color="red")

# ── Fill between: shade a region ─────────────────────────────────────────
ax.fill_between(x, y, 0, where=(y > 0), alpha=0.15, color="steelblue", label="positive area")
ax.fill_between(x, y, 0, where=(y < 0), alpha=0.15, color="salmon",    label="negative area")

ax.set_xlim(0, 2*np.pi)
ax.set_title("Annotations, Reference Lines & Fill")
ax.legend(loc="upper right", fontsize=9)
ax.grid(True, linestyle=":", alpha=0.4)

plt.tight_layout()
plt.show()
```

## 11. Styling & Themes

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 200)

# ── Available styles ──────────────────────────────────────────────────────
print("Built-in styles:", plt.style.available[:10], "...")

# ── Show the same plot in 4 different styles ──────────────────────────────
styles = ["seaborn-v0_8-darkgrid", "ggplot", "dark_background", "bmh"]

fig, axes = plt.subplots(1, 4, figsize=(18, 4))

for ax, style in zip(axes, styles):
    with plt.style.context(style):           # apply style temporarily
        ax.plot(x, np.sin(x), linewidth=2)
        ax.plot(x, np.cos(x), linewidth=2)
        ax.set_title(style, fontsize=9)
        ax.set_xticks([])

plt.suptitle("4 Built-in Styles", fontsize=13)
plt.tight_layout()
plt.show()

# ── rcParams: global settings ────────────────────────────────────────────
plt.rcParams.update({
    "font.size":       12,
    "axes.linewidth":  1.2,
    "lines.linewidth": 2,
    "figure.dpi":      100,
})

# ── Custom color cycle ────────────────────────────────────────────────────
from cycler import cycler
my_colors = ["#264653", "#2a9d8f", "#e9c46a", "#f4a261", "#e76f51"]
plt.rcParams["axes.prop_cycle"] = cycler("color", my_colors)

fig, ax = plt.subplots(figsize=(8, 4))
for i in range(5):
    ax.plot(x, np.sin(x + i * np.pi/5), label=f"shift {i}")
ax.set_title("Custom color cycle via rcParams")
ax.legend()
plt.tight_layout()
plt.show()

# ── Reset to defaults ─────────────────────────────────────────────────────
plt.rcParams.update(plt.rcParamsDefault)
```

## 12. Colormaps & Images

```python
import matplotlib.pyplot as plt
import numpy as np

# ── imshow(): display a 2-D array as an image ─────────────────────────────
x = np.linspace(-3, 3, 200)
y = np.linspace(-3, 3, 200)
X, Y = np.meshgrid(x, y)
Z    = np.sin(X) * np.cos(Y)          # 2-D function

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# ── Different colormaps ────────────────────────────────────────────────────
cmaps = ["viridis", "RdBu_r", "plasma"]
for ax, cmap in zip(axes, cmaps):
    im = ax.imshow(Z, cmap=cmap, origin="lower",
                   extent=[-3, 3, -3, 3],    # set axis scale
                   aspect="auto")
    fig.colorbar(im, ax=ax, shrink=0.85)
    ax.set_title(f"cmap='{cmap}'")
    ax.set_xlabel("x")
    ax.set_ylabel("y")

plt.suptitle("imshow — sin(x)·cos(y)", fontsize=13)
plt.tight_layout()
plt.show()

# ── matshow(): visualize a matrix (e.g. correlation) ─────────────────────
import numpy as np

corr = np.corrcoef(np.random.randn(6, 100))   # 6×6 correlation matrix
labels = [f"Feat {i}" for i in range(6)]

fig2, ax2 = plt.subplots(figsize=(5, 4))
ms = ax2.matshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
fig2.colorbar(ms)
ax2.set_xticks(range(6)); ax2.set_xticklabels(labels, rotation=45, ha="left")
ax2.set_yticks(range(6)); ax2.set_yticklabels(labels)
ax2.set_title("Correlation Matrix", pad=20)

# Add numeric values inside cells
for (i, j), val in np.ndenumerate(corr):
    ax2.text(j, i, f"{val:.2f}", ha="center", va="center", fontsize=7,
             color="white" if abs(val) > 0.5 else "black")

plt.tight_layout()
plt.show()
```

## 13. 3D Plots

```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D   # registers the 3D projection

fig = plt.figure(figsize=(16, 5))

# ── 1. Surface plot ───────────────────────────────────────────────────────
ax1 = fig.add_subplot(131, projection="3d")

x = np.linspace(-3, 3, 60)
y = np.linspace(-3, 3, 60)
X, Y = np.meshgrid(x, y)
Z    = np.sin(np.sqrt(X**2 + Y**2))

surf = ax1.plot_surface(X, Y, Z, cmap="viridis", alpha=0.9)
fig.colorbar(surf, ax=ax1, shrink=0.5)
ax1.set_title("Surface Plot")
ax1.set_xlabel("X"); ax1.set_ylabel("Y"); ax1.set_zlabel("Z")

# ── 2. Wireframe plot ────────────────────────────────────────────────────
ax2 = fig.add_subplot(132, projection="3d")
ax2.plot_wireframe(X, Y, Z, color="steelblue", linewidth=0.5, rstride=5, cstride=5)
ax2.set_title("Wireframe Plot")
ax2.set_xlabel("X"); ax2.set_ylabel("Y"); ax2.set_zlabel("Z")

# ── 3. 3D Scatter ─────────────────────────────────────────────────────────
ax3  = fig.add_subplot(133, projection="3d")
rng  = np.random.default_rng(42)
n    = 200
xs   = rng.normal(0, 1, n)
ys   = rng.normal(0, 1, n)
zs   = xs**2 + ys**2 + rng.normal(0, 0.5, n)

sc = ax3.scatter(xs, ys, zs, c=zs, cmap="plasma", s=20, alpha=0.7)
fig.colorbar(sc, ax=ax3, shrink=0.5)
ax3.set_title("3D Scatter")
ax3.set_xlabel("X"); ax3.set_ylabel("Y"); ax3.set_zlabel("Z")

plt.suptitle("3D Plots with mpl_toolkits.mplot3d", fontsize=13)
plt.tight_layout()
plt.show()
```

## 14. Saving Figures

```python
import matplotlib.pyplot as plt
import numpy as np
import os

x = np.linspace(0, 2 * np.pi, 200)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, np.sin(x), color="steelblue", linewidth=2)
ax.set_title("Figure saved in multiple formats")
ax.set_xlabel("x"); ax.set_ylabel("sin(x)")
ax.grid(True, linestyle=":", alpha=0.5)
plt.tight_layout()

# ── PNG: raster, good for web / presentations ─────────────────────────────
fig.savefig("plot.png", dpi=150, bbox_inches="tight")

# ── PDF: vector, best for papers / LaTeX ─────────────────────────────────
fig.savefig("plot.pdf", bbox_inches="tight")

# ── SVG: vector, editable in Inkscape / Illustrator ──────────────────────
fig.savefig("plot.svg", bbox_inches="tight")

# ── JPEG: smaller size, slight quality loss ───────────────────────────────
fig.savefig("plot.jpg", dpi=150, bbox_inches="tight", quality=90)

plt.show()

# ── Compare file sizes ────────────────────────────────────────────────────
for fname in ["plot.png", "plot.pdf", "plot.svg", "plot.jpg"]:
    size = os.path.getsize(fname)
    print(f"  {fname:12s}: {size:>7,} bytes")

# ── Key savefig parameters ────────────────────────────────────────────────
# dpi          → dots per inch (72=screen, 150=good, 300=print quality)
# bbox_inches  → "tight" removes whitespace padding around the figure
# transparent  → True makes background transparent (great for overlays)
# facecolor    → set background color  (e.g. facecolor="white")
# format       → force format even if extension differs
```

## 15. Practical ML Plots

```python
import matplotlib.pyplot as plt
import numpy as np

# ══════════════════════════════════════════════════════════════════════════
# 1. Training & Validation Loss Curve
# ══════════════════════════════════════════════════════════════════════════
epochs     = np.arange(1, 51)
train_loss = 1.2 * np.exp(-0.08 * epochs) + 0.05 + np.random.default_rng(0).normal(0, 0.01, 50)
val_loss   = 1.2 * np.exp(-0.06 * epochs) + 0.12 + np.random.default_rng(1).normal(0, 0.015, 50)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(epochs, train_loss, label="Train loss", color="steelblue",  linewidth=2)
ax.plot(epochs, val_loss,   label="Val loss",   color="#e76f51",    linewidth=2, linestyle="--")
ax.axvline(35, color="gray", linestyle=":", linewidth=1.5, label="Early stop")
ax.set_title("Training & Validation Loss")
ax.set_xlabel("Epoch"); ax.set_ylabel("Loss")
ax.legend(); ax.grid(True, linestyle=":", alpha=0.5)
plt.tight_layout(); plt.show()

# ══════════════════════════════════════════════════════════════════════════
# 2. Confusion Matrix
# ══════════════════════════════════════════════════════════════════════════
cm     = np.array([[50, 3, 2],
                   [ 4, 45, 6],
                   [ 1,  5, 48]])
labels = ["Cat", "Dog", "Bird"]

fig2, ax2 = plt.subplots(figsize=(5, 4))
im = ax2.matshow(cm, cmap="Blues")
fig2.colorbar(im)
ax2.set_xticks(range(3)); ax2.set_xticklabels(labels)
ax2.set_yticks(range(3)); ax2.set_yticklabels(labels)
ax2.set_xlabel("Predicted"); ax2.set_ylabel("Actual")
ax2.set_title("Confusion Matrix", pad=15)

for (i, j), val in np.ndenumerate(cm):
    ax2.text(j, i, str(val), ha="center", va="center",
             fontsize=13, color="white" if val > 30 else "black")
plt.tight_layout(); plt.show()

# ══════════════════════════════════════════════════════════════════════════
# 3. Feature Importance Bar Chart
# ══════════════════════════════════════════════════════════════════════════
features   = ["Age", "Salary", "Experience", "Education", "City", "Dept"]
importance = np.array([0.32, 0.25, 0.18, 0.12, 0.08, 0.05])
order      = np.argsort(importance)   # sort ascending for horizontal bar

fig3, ax3 = plt.subplots(figsize=(7, 4))
bars = ax3.barh([features[i] for i in order],
                importance[order],
                color=plt.cm.viridis(importance[order] / importance.max()))
ax3.set_xlabel("Importance Score")
ax3.set_title("Feature Importance")
ax3.spines["top"].set_visible(False)
ax3.spines["right"].set_visible(False)

for bar, val in zip(bars, importance[order]):
    ax3.text(val + 0.005, bar.get_y() + bar.get_height()/2,
             f"{val:.2f}", va="center", fontsize=10)

plt.tight_layout(); plt.show()
```
