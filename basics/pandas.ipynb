## 1. Import & Data Structures

```python
import pandas as pd
import numpy as np

# ── Series: 1-D labeled array (like a column in a spreadsheet) ────────────
s = pd.Series([10, 20, 30, 40, 50])
print("Series:\n", s)
print("Values:", s.values)       # underlying NumPy array
print("Index: ", s.index)        # RangeIndex(start=0, stop=5, step=1)

# ── Series with custom labels ──────────────────────────────────────────────
cities = pd.Series([3.5, 8.9, 13.2], index=["Tehran", "London", "Tokyo"])
print("\nCities Series:\n", cities)
print("Tehran value:", cities["Tehran"])   # access by label

# ── DataFrame: 2-D labeled table (rows × columns) ─────────────────────────
df = pd.DataFrame({
    "name":  ["Alice", "Bob", "Carol"],
    "age":   [25, 30, 27],
    "score": [88.5, 92.0, 79.3],
})
print("\nDataFrame:\n", df)
```

## 4. Indexing & Selection

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,salary,city
Alice,28,55000,Tehran
Bob,35,72000,London
Carol,24,48000,Tokyo
Dave,42,95000,Tehran
Eve,31,61000,London
"""
df = pd.read_csv(StringIO(csv_text))

# ── Select a single column → returns a Series ─────────────────────────────
print("Single column:\n", df["name"])

# ── Select multiple columns → returns a DataFrame ─────────────────────────
print("\nMultiple columns:\n", df[["name", "salary"]])

# ── loc: label-based selection [rows, columns] ────────────────────────────
print("\nloc rows 0-2, cols name+age:\n", df.loc[0:2, ["name", "age"]])

# ── iloc: integer position-based selection ────────────────────────────────
print("\niloc first 3 rows, first 2 cols:\n", df.iloc[:3, :2])

# ── Single cell ───────────────────────────────────────────────────────────
print("\nCell at row 1, col 'salary':", df.loc[1, "salary"])   # 72000
print("Cell at position [2, 0]:    ", df.iloc[2, 0])          # Carol
```

## 5. Filtering & Querying

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,salary,city
Alice,28,55000,Tehran
Bob,35,72000,London
Carol,24,48000,Tokyo
Dave,42,95000,Tehran
Eve,31,61000,London
"""
df = pd.read_csv(StringIO(csv_text))

# ── Boolean mask ──────────────────────────────────────────────────────────
print("Age > 30:\n", df[df["age"] > 30])

# ── Combine conditions with & (and) | (or) ────────────────────────────────
print("\nAge > 25 AND city is London:\n",
      df[(df["age"] > 25) & (df["city"] == "London")])

# ── query(): SQL-like string syntax ───────────────────────────────────────
print("\nquery salary > 60000:\n", df.query("salary > 60000"))

# ── isin(): match against a list of values ────────────────────────────────
print("\nCity in Tehran or Tokyo:\n", df[df["city"].isin(["Tehran", "Tokyo"])])

# ── between(): range filter (inclusive) ──────────────────────────────────
print("\nAge between 28 and 35:\n", df[df["age"].between(28, 35)])
```

## 6. Adding & Modifying Columns

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,salary,city
Alice,28,55000,Tehran
Bob,35,72000,London
Carol,24,48000,Tokyo
Dave,42,95000,Tehran
Eve,31,61000,London
"""
df = pd.read_csv(StringIO(csv_text))

# ── Direct assignment ─────────────────────────────────────────────────────
df["salary_k"] = df["salary"] / 1000      # salary in thousands
df["senior"]   = df["age"] >= 35           # boolean column

# ── apply(): custom function on each element ──────────────────────────────
df["level"] = df["salary"].apply(lambda x: "high" if x > 60000 else "normal")
print("With apply():\n", df[["name", "salary", "level"]])

# ── map(): element-wise mapping via dict ─────────────────────────────────
city_region = {"Tehran": "Asia", "London": "Europe", "Tokyo": "Asia"}
df["region"] = df["city"].map(city_region)
print("\nWith map():\n", df[["name", "city", "region"]])

# ── Rename columns ────────────────────────────────────────────────────────
df.rename(columns={"salary_k": "sal_k"}, inplace=True)
print("\nRenamed columns:", df.columns.tolist())

# ── Drop a column ─────────────────────────────────────────────────────────
df.drop(columns=["senior"], inplace=True)
print("After drop:\n", df.head())
```

## 7. Handling Missing Data

```python
import pandas as pd
import numpy as np

# ── DataFrame with missing values ─────────────────────────────────────────
df = pd.DataFrame({
    "name":   ["Alice", "Bob", "Carol", "Dave", "Eve"],
    "age":    [28, np.nan, 24, 42, np.nan],
    "salary": [55000, 72000, np.nan, 95000, 61000],
    "city":   ["Tehran", "London", None, "Tehran", "London"],
})
print("DataFrame with NaN:\n", df)

# ── Detect missing values ─────────────────────────────────────────────────
print("\nMissing count per column:\n", df.isnull().sum())

# ── Drop rows with any NaN ────────────────────────────────────────────────
print("\ndropna():\n", df.dropna())

# ── Fill NaN with specific values per column ─────────────────────────────
df_filled = df.fillna({"age": df["age"].mean(), "salary": 0, "city": "Unknown"})
print("\nfillna with mean/defaults:\n", df_filled)

# ── Forward fill: propagate last valid value forward ──────────────────────
print("\nffill:\n", df.ffill())

# ── Interpolate numeric columns linearly ──────────────────────────────────
print("\ninterpolate():\n", df[["age", "salary"]].interpolate())
```

## 8. Statistical & Aggregation

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,salary,city
Alice,28,55000,Tehran
Bob,35,72000,London
Carol,24,48000,Tokyo
Dave,42,95000,Tehran
Eve,31,61000,London
"""
df = pd.read_csv(StringIO(csv_text))

# ── Basic aggregations ────────────────────────────────────────────────────
print("Mean salary: ", df["salary"].mean())
print("Max age:     ", df["age"].max())
print("Sum salaries:", df["salary"].sum())
print("Median age:  ", df["age"].median())
print("Std salary:  ", df["salary"].std().round(2))

# ── value_counts(): frequency of each unique value ───────────────────────
print("\nCity counts:\n", df["city"].value_counts())

# ── Correlation matrix ────────────────────────────────────────────────────
print("\nCorrelation matrix:\n", df[["age", "salary"]].corr())

# ── nunique(): number of distinct values ─────────────────────────────────
print("\nUnique cities:", df["city"].nunique())

# ── agg on whole DataFrame ───────────────────────────────────────────────
print("\nMultiple stats at once:\n",
      df[["age", "salary"]].agg(["mean", "min", "max", "std"]))
```

## 9. Sorting & Ranking

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,salary,city
Alice,28,55000,Tehran
Bob,35,72000,London
Carol,24,48000,Tokyo
Dave,42,95000,Tehran
Eve,31,61000,London
"""
df = pd.read_csv(StringIO(csv_text))

# ── sort_values(): sort by one column ────────────────────────────────────
print("Sorted by salary (desc):\n", df.sort_values("salary", ascending=False))

# ── Sort by multiple columns ──────────────────────────────────────────────
print("\nSorted by city then age:\n",
      df.sort_values(["city", "age"]))

# ── nlargest / nsmallest ──────────────────────────────────────────────────
print("\nTop 3 salaries:\n", df.nlargest(3, "salary"))
print("\nYoungest 2:\n",     df.nsmallest(2, "age"))

# ── rank(): assign a rank to each value ──────────────────────────────────
df["salary_rank"] = df["salary"].rank(ascending=False).astype(int)
print("\nWith salary rank:\n", df[["name", "salary", "salary_rank"]])
```

## 10. GroupBy & Aggregation

```python
import pandas as pd
from io import StringIO

csv_text = """name,age,salary,city
Alice,28,55000,Tehran
Bob,35,72000,London
Carol,24,48000,Tokyo
Dave,42,95000,Tehran
Eve,31,61000,London
"""
df = pd.read_csv(StringIO(csv_text))

# ── groupby + single aggregation ─────────────────────────────────────────
print("Average salary per city:\n",
      df.groupby("city")["salary"].mean())

# ── groupby + multiple aggregations ──────────────────────────────────────
print("\nMultiple agg per city:\n",
      df.groupby("city").agg(
          avg_salary=("salary", "mean"),
          max_age   =("age",    "max"),
          count     =("name",   "count")
      ))

# ── transform(): broadcast group result back to original index ────────────
df["city_avg_salary"] = df.groupby("city")["salary"].transform("mean")
print("\nWith city avg salary column:\n",
      df[["name", "city", "salary", "city_avg_salary"]])

# ── pivot_table(): Excel-style pivot ──────────────────────────────────────
df["dept"] = ["Eng", "HR", "Eng", "HR", "Eng"]
pivot = df.pivot_table(values="salary", index="city", columns="dept", aggfunc="mean")
print("\nPivot table:\n", pivot)
```

## 11. Merging & Joining

```python
import pandas as pd

employees = pd.DataFrame({
    "emp_id":  [1, 2, 3, 4],
    "name":    ["Alice", "Bob", "Carol", "Dave"],
    "dept_id": [10, 20, 10, 30],
})

departments = pd.DataFrame({
    "dept_id":   [10, 20, 30],
    "dept_name": ["Engineering", "HR", "Finance"],
})

# ── merge(): like SQL JOIN ────────────────────────────────────────────────
inner = pd.merge(employees, departments, on="dept_id", how="inner")
print("Inner merge:\n", inner)

# ── left join: keep all employees even without a matching dept ────────────
extra = pd.DataFrame({"emp_id": [5], "name": ["Eve"], "dept_id": [99]})
all_emp = pd.concat([employees, extra], ignore_index=True)
left = pd.merge(all_emp, departments, on="dept_id", how="left")
print("\nLeft join (Eve has no dept):\n", left)

# ── concat(): stack DataFrames vertically ────────────────────────────────
df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"A": [5, 6], "B": [7, 8]})
stacked = pd.concat([df1, df2], ignore_index=True)
print("\nVertical concat:\n", stacked)

# ── concat() side-by-side (axis=1) ───────────────────────────────────────
side = pd.concat([df1, df2], axis=1)
print("\nHorizontal concat:\n", side)
```

## 12. String Operations

```python
import pandas as pd

df = pd.DataFrame({
    "email": [
        "Alice@Gmail.COM",
        "bob.smith@yahoo.com",
        "carol_99@hotmail.com",
        "DAVE@Work.org",
    ]
})

# ── str accessor: element-wise string methods ─────────────────────────────
df["lower"]  = df["email"].str.lower()
df["length"] = df["email"].str.len()

# ── contains(): boolean filter ────────────────────────────────────────────
print("Gmail users:\n",
      df[df["email"].str.lower().str.contains("gmail")])

# ── split(): extract part of a string ────────────────────────────────────
df["domain"] = df["email"].str.lower().str.split("@").str[1]
print("\nExtracted domain:\n", df[["email", "domain"]])

# ── replace(): regex-powered replace ─────────────────────────────────────
df["clean"] = df["email"].str.replace(r"[^a-zA-Z0-9@.]", "", regex=True)
print("\nCleaned email:\n", df[["email", "clean"]])

# ── startswith / endswith ─────────────────────────────────────────────────
print("\nEnds with .com:\n",
      df[df["email"].str.lower().str.endswith(".com")])
```

## 13. DateTime Operations

```python
import pandas as pd
import numpy as np

# ── Parse strings to datetime ─────────────────────────────────────────────
df = pd.DataFrame({
    "date":  ["2024-01-15", "2024-02-20", "2024-03-10", "2024-04-05"],
    "sales": [1200, 1800, 1500, 2200],
})
df["date"] = pd.to_datetime(df["date"])
print("dtype after parsing:", df["date"].dtype)

# ── dt accessor: extract date parts ──────────────────────────────────────
df["year"]    = df["date"].dt.year
df["month"]   = df["date"].dt.month
df["day"]     = df["date"].dt.day
df["weekday"] = df["date"].dt.day_name()
print("\nDate parts:\n", df)

# ── Date arithmetic ───────────────────────────────────────────────────────
df["next_week"] = df["date"] + pd.Timedelta(days=7)
print("\nNext week:\n", df[["date", "next_week"]])

# ── date_range(): generate a sequence of dates ───────────────────────────
monthly = pd.date_range(start="2024-01-01", periods=6, freq="MS")
print("\nMonthly date range:\n", monthly)

# ── resample(): aggregate time-series data by period ─────────────────────
ts = pd.Series(
    np.random.randint(100, 500, 90),
    index=pd.date_range("2024-01-01", periods=90, freq="D")
)
print("\nMonthly totals (resample):\n", ts.resample("ME").sum())
```

## 14. Reshaping Data

```python
import pandas as pd

# ── Wide format ───────────────────────────────────────────────────────────
df_wide = pd.DataFrame({
    "city":  ["Tehran", "London", "Tokyo"],
    "jan":   [5, 4, 6],
    "feb":   [7, 5, 8],
    "mar":   [12, 8, 10],
})
print("Wide format:\n", df_wide)

# ── melt(): wide → long (tidy format) ────────────────────────────────────
df_long = df_wide.melt(
    id_vars="city",
    value_vars=["jan", "feb", "mar"],
    var_name="month",
    value_name="temp"
)
print("\nLong (melted) format:\n", df_long)

# ── pivot(): long → wide ──────────────────────────────────────────────────
df_back = df_long.pivot(index="city", columns="month", values="temp")
print("\nPivoted back to wide:\n", df_back)

# ── stack(): move column labels into the row index ────────────────────────
stacked = df_back.stack()
print("\nStacked (MultiIndex Series):\n", stacked.head(6))

# ── unstack(): move innermost row index back to columns ───────────────────
print("\nUnstacked back:\n", stacked.unstack())
```

```python
import pandas as pd

df = pd.DataFrame({
    "name":   ["Alice", "Bob", "Carol", "Dave"],
    "age":    [28, 35, 24, 42],
    "salary": [55000, 72000, 48000, 95000],
    "city":   ["Tehran", "London", "Tokyo", "Tehran"],
})

# ═══════════════════════════════════════════════════════════════════════════
# 4. Parquet  ──────────────────────────────────────────────────────────────
# ═══════════════════════════════════════════════════════════════════════════
# Requires:  pip install pyarrow   (or fastparquet)
# Best for:  large datasets, columnar storage, fast reads, small file size

# ── Write to Parquet ──────────────────────────────────────────────────────
df.to_parquet("employees.parquet", index=False)
print("Saved to employees.parquet")

# ── Read back ─────────────────────────────────────────────────────────────
df_pq = pd.read_parquet("employees.parquet")
print("\nRead from Parquet:\n", df_pq)

# ── Read only specific columns (columnar advantage) ───────────────────────
df_cols = pd.read_parquet("employees.parquet", columns=["name", "salary"])
print("\nPartial read (name + salary):\n", df_cols)

# ── Compression options ───────────────────────────────────────────────────
df.to_parquet("employees_snappy.parquet", compression="snappy")   # fast
df.to_parquet("employees_gzip.parquet",   compression="gzip")     # smaller

import os
sizes = {
    "CSV":            os.path.getsize("employees.csv"),
    "Parquet(snappy)":os.path.getsize("employees_snappy.parquet"),
    "Parquet(gzip)":  os.path.getsize("employees_gzip.parquet"),
}
print("\nFile sizes (bytes):")
for fmt, size in sizes.items():
    print(f"  {fmt:20s}: {size}")
```

```python
import pandas as pd

df = pd.DataFrame({
    "name":   ["Alice", "Bob", "Carol", "Dave"],
    "age":    [28, 35, 24, 42],
    "salary": [55000, 72000, 48000, 95000],
    "city":   ["Tehran", "London", "Tokyo", "Tehran"],
})

# ═══════════════════════════════════════════════════════════════════════════
# 5. Pickle  ───────────────────────────────────────────────────────────────
# ═══════════════════════════════════════════════════════════════════════════
# Best for:  saving Python objects exactly (dtypes, index, categoricals)
# Warning:   not safe to load from untrusted sources

df.to_pickle("employees.pkl")
df_pkl = pd.read_pickle("employees.pkl")
print("Pickle round-trip:\n", df_pkl)

# ═══════════════════════════════════════════════════════════════════════════
# 6. HTML table  ───────────────────────────────────────────────────────────
# ═══════════════════════════════════════════════════════════════════════════
# Requires:  pip install lxml  (or html5lib / beautifulsoup4)

# ── Write DataFrame as an HTML table ─────────────────────────────────────
html_str = df.to_html(index=False)
print("\nHTML snippet (first 300 chars):\n", html_str[:300])

# ── Read tables from an HTML string / URL ────────────────────────────────
tables = pd.read_html(html_str)          # returns a LIST of DataFrames
print("\nRead from HTML:\n", tables[0])

# ── Real-world use: scrape a table from a web page ───────────────────────
# tables = pd.read_html("https://en.wikipedia.org/wiki/List_of_countries_by_GDP")
# gdp_df = tables[0]

# ═══════════════════════════════════════════════════════════════════════════
# 7. SQLite (SQL database)  ────────────────────────────────────────────────
# ═══════════════════════════════════════════════════════════════════════════
import sqlite3

conn = sqlite3.connect(":memory:")            # in-memory DB (no file needed)

# ── Write DataFrame to SQL table ─────────────────────────────────────────
df.to_sql("employees", conn, if_exists="replace", index=False)
print("\nSaved to SQLite table 'employees'")

# ── Read with a SQL query ─────────────────────────────────────────────────
df_sql = pd.read_sql("SELECT * FROM employees WHERE salary > 60000", conn)
print("\nSQL query result:\n", df_sql)

# ── read_sql_query vs read_sql_table ──────────────────────────────────────
df_tbl = pd.read_sql_table("employees", conn)   # read entire table
print("\nFull table:\n", df_tbl)

conn.close()
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    "age":    np.random.randint(20, 60, 10),
    "salary": np.random.randint(30000, 100000, 10),
    "city":   np.random.choice(["Tehran", "London", "Tokyo"], 10),
    "churn":  np.random.randint(0, 2, 10),   # target: 0 or 1
})
print("Raw data:\n", df)

# ── 1. One-hot encode categorical column ──────────────────────────────────
df_enc = pd.get_dummies(df, columns=["city"], drop_first=False)
print("\nAfter one-hot encoding:\n", df_enc)

# ── 2. Min-Max normalization ──────────────────────────────────────────────
for col in ["age", "salary"]:
    col_min, col_max = df[col].min(), df[col].max()
    df_enc[col] = (df[col] - col_min) / (col_max - col_min)
print("\nNormalized age & salary:\n", df_enc[["age", "salary"]])

# ── 3. Separate features and target ──────────────────────────────────────
X = df_enc.drop(columns="churn")
y = df_enc["churn"]
print("\nFeature matrix X shape:", X.shape)
print("Target y shape:        ", y.shape)

# ── 4. Train / test split (80 / 20) ──────────────────────────────────────
split    = int(0.8 * len(df_enc))
X_train  = X.iloc[:split]
X_test   = X.iloc[split:]
y_train  = y.iloc[:split]
y_test   = y.iloc[split:]
print(f"\nTrain: {len(X_train)} rows | Test: {len(X_test)} rows")
print("X_train:\n", X_train)
```
